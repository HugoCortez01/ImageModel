### `4.` Criar um programa para somar os números inteiros de um intervalo entre **`i`** (início) e **`f`** (fim) (inclusive i e f), com **i** e **f** fornecidos pelo usuário.
+ ### Contador e Acumulador

```python
# usando o while

i = int(input('Início: '))
f = int(input('\nFim: '))

c = i  # c é o contador e i é o início do intervalo
s = 0  # s é o acumulador (0 é o elemento neutro da soma)
while c <= f:  # f é o fim do intervalo
    s += c  # s = s + c (s é o acumulador)
    c += 1  # incrementa o contador

print(f'\nA soma entre {i} e {f} é igual a {s}')
```

### `Outra forma`: usando o for.

```python
# usando o for

i = int(input('Início: '))
f = int(input('\nFim: '))

s = 0  # soma (acumulador)
for c in range(i, f + 1):
    s += c  # s = s + c

print(f'\nA soma entre {i} e {f} é igual a {s}')
```

### `5.` Criar uma programa para ler uma certa qtde de valores reais (`float`) e, após as leituras, mostrar a média aritmética dos números digitados. A qtde de valores é escolhida pelo usuário no início do programa.
### Ex.:
    Qtde de valores: 5
    Valor 1: 6
    Valor 2: 2
    Valor 3: 7
    Valor 4: 5
    Valor 5: 1
    Média aritmética = 4.20

```python
qv = int(input('Qtde de valores: '))
print()  # salta uma linha
s = 0  # armazena a soma dos valores (acumulador)
for i in range(qv):  # repete qv vezes
    val = float(input(f'Valor {i + 1}: '))  # leitura dos valores
    s += val  # soma dos valores

ma = s / qv  # cálculo da média
print(f'\nMédia aritmética = {ma:.2f}')
```

### `2.` Gerar uma lista com 20 valores aleatórios entre 10 e 20 e, em seguida, ordenar a lista, calcular e mostrar (em ordem crescente) a qtde de ocorrências de cada valor dentro da lista (não repetir valores já mostrados).

```python
from random import randint

lis = [randint(10, 20) for _ in range(20)]
lis.sort()
print(f'Lista: {lis}\n')

print('Valores e quantidades:\n')
sel = []  # lista de valores já selecionados (mostrados)
for v in lis:
    if v not in sel:
        cont = 0
        for x in lis:
            if x == v:
                cont += 1
        print(f'{v}: {cont}x')
        sel.append(v)

print('\nValores e quantidades:\n')
sel = []  # lista de valores já selecionados (mostrados)
for v in lis:
    if v not in sel:
        print(f'{v}: {lis.count(v)}x')
        sel.append(v)

# Outra forma
print('\nValores e quantidades:\n')
for v in sorted(set(lis)):  # set (conjunto) elimina repetições em lis
    print(f'{v}: {lis.count(v)}x')
```

Após a criação das listas, ordená-las em ordem crescente e criar uma terceira lista com os valores presentes em ambas as listas (1ª e 2ª), ou seja, presente tanto na primeira como na segunda lista. Mostrar as 3 listas no final do programa.

```python
from random import sample

lis1 = sample(range(1, 101), 10)  # lista com 10 números distintos entre 1 e 100
lis1.sort()
lis2 = sample(range(1, 101), 20)  # lista com 20 números
lis2.sort()

lis3 = []
for v in lis1:
    if v in lis2:
        lis3.append(v)

print(f'Lista 1: {lis1}')
print(f'\nLista 2: {lis2}')
print(f'\nLista 3 = {lis3}')
```

```python
# Mostrar uma matriz linha por linha

mat = [[9] * 3 for i in range(5)]

print('Matriz:')
for i in range(5):
    print(mat[i])  # i é a linha

print('\nMatriz:')
for lin in mat:  # mostra a matriz linha por linha
    print(lin)

print('\nMatriz:')
for i in range(5):  # i é linha
    for j in range(3):  # j é coluna
        print(mat[i][j], end=' ')
    print()

print('\nMatriz:')
for lin in mat:  # lin recebe uma linha
    for v in lin:  # v recebe valor por valor da linha (lin)
        print(v, end=' ')
    print()
```

```python
from random import randint

def main():
    mat = [[randint(0, 9) for j in range(5)] for i in range(5)]
    print('Matriz:')
    for lin in mat:
        print(lin)

    s = 0
    for i in range(5):
        s += mat[i][i]
    print(f'\nSoma da diagonal principal = {s}')

    # p = 1
    # j = 4
    # for i in range(5):
    #     p *= mat[i][j]
    #     j -= 1

    p = 1
    for i in range(5):
        p *= mat[i][4 - i]
    print(f'\nProduto da diagonal secundária = {p}')

main()
```

### `3.` Gerar e mostrar uma matriz 3x4 de inteiros aleatórios entre 10 e 30 e, em seguida, calcular e mostrar a qtde de pares e a qtde de ímpares contidos nessa matriz.

```python
from random import randint
import numpy as np

def main():
    mat = [[randint(10, 30) for j in range(4)] for i in range(3)]
    print('Matriz:')
    for lin in mat:
        print(lin)

    qp = qi = 0
    for lin in mat:
        for v in lin:
            if v % 2 == 0:
                qp += 1
            else:
                qi += 1

    npmat = np.array(mat)
    qp = np.count_nonzero(npmat % 2 == 0)  # ou: npmat[npmat % 2 == 0].size
    qi = npmat.size - qp

    print(f'\nQtde de pares = {qp}')
    print(f'\nQtde de ímpares = {qi}')

main()
```

### `4.` Gerar e mostrar uma matriz 4x5 com números inteiros e aleatórios entre 0 e 9 e, em seguida, calcular e mostrar:
+ ### O menor e o maior valor de cada linha.
+ ### A média aritmética de cada linha e inseri-la na 6ª coluna. Mostrar a matriz após o cálculo.

```python
from random import randint

def main():
    mat = [[randint(0, 9) for j in range(5)] for i in range(4)]
    print('Matriz:')
    for lin in mat:
        print(lin)

    for i in range(4):
        print(f'\n{i + 1}ª linha:\nMenor = {min(mat[i])}\nMaior = {max(mat[i])}')

    for lin in mat:
        lin.append(sum(lin) / len(lin))  # ou dividir por 5

    print('\nMatriz com as médias aritméticas de cada linha:')
    for lin in mat:
        print(lin)

main()
```

```python
from random import randint

def main():
    mat = [[randint(0, 9) for j in range(3)] for i in range(5)]
    print('Matriz:')
    for lin in mat:
        print(lin)

    medcol = [] # lista com as médias das colunas
    for j in range(3):  # percorre a matriz coluna por coluna
        col = []
        for i in range(5): # linha
            col.append(mat[i][j])
        print(f'\n{j + 1}ª coluna:\nMenor = {min(col)}\nMaior = {max(col)}')
        medcol.append(sum(col) / len(col))  # ou dividir por 3
    mat.append(medcol)

    print('\nMatriz:')
    for lin in mat:
        print(lin)

main()
```

```python
# usando o módulo (biblioteca) Numpy

from random import randint
import numpy as np

def main():
    mat = [[randint(1, 10) for j in range(3)] for i in range(5)]
    npmat = np.array(mat)

    print('Matriz:')
    print(npmat)

    medcol = []
    for j in range(3):  # coluna
        print(f'\n{j + 1}ª coluna:\nMenor = {npmat[:, j].min()}\nMaior = {npmat[:, j].max()}')
        medcol.append(np.average(npmat[:, j]))

    # inserir uma linha em um array Numpy
    npmat = np.append(npmat, [medcol], axis=0)
    print('\nMatriz com as médias aritméticas de cada coluna:')
    print(npmat)

main()
```

### **Sem** parâmetros e **sem** retorno:

```python
def msg():
    print('Ciência da Computação - 2021')

# programa principal

msg()
```

### **Sem** parâmetros e **com** retorno:

```python
def soma():
    a = int(input('A: '))
    b = int(input('\nB: '))
    return a + b

# programa principal

s = soma()
print(f'\nSoma = {s}')
if soma() > 10:
    print('Maior que 10')
```

### **Com** parâmetros e **sem** retorno:

```python
def potencia(base, exp): # pbase e pexp são parâmetros (dados externos passados à função)
    p = base ** exp
    print(f'{base} elevado a {exp} = {p}')

# programa principal

potencia(2, 10)
potencia(3, 5)
b = 5
e = 8
potencia(b, e)
```

### **Com** parâmetros e **com** retorno:

```python
def menor3(a, b, c):
    return min(a, b, c)

# programa principal

def main():
    va = int(input('A: '))
    vb = int(input('\nB: '))
    vc = int(input('\nC: '))
    me = menor3(va, vb, vc)
    print(f'\nMenor = {me}')

    if menor3(va, vb, vc) == va:
        print('\nOK')

main()
```
